# Simulated Datasets (Python Version)

This notebook contains the Python equivalents of the simulated datasets originally created in R.

Each section simulates a different dataset using `numpy` and `pandas`. The random seed (`np.random.seed(42)`) ensures reproducibility — this means everyone will get the same results, which is important for grading or collaborative exercises.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Set seed for reproducibility
np.random.seed(42)
```

## 1. Global Temperatures Over Time

**Description:** Annual global average temperature (in degrees Celsius) for the years 1900 to 2020.

**Columns:**
- Year: (1900–2020)
- Global_Temperature: Average global temperature each year.

```python
years = np.arange(1900, 2021)
global_temp = np.round(14 + np.random.normal(loc=0.05, scale=0.2, size=len(years)), 2)

global_temp_data = pd.DataFrame({
    'Year': years,
    'Global_Temperature': global_temp
})
global_temp_data

plt.figure(figsize=(10,4))
plt.plot(global_temp_data['Year'], global_temp_data['Global_Temperature'], color='tab:blue')
plt.xlabel('Year')
plt.ylabel('Temperature (°C)')
plt.title('Global Average Temperature (1900-2020)')
plt.grid(True)
plt.tight_layout()
plt.show()
```

## 2. Impact of Advertising Budget on Sales Revenue

**Description:** Relationship between a company's monthly advertising budget (in thousands of dollars) and its corresponding monthly sales revenue (in thousands of dollars) over a two-year period.

**Columns:**
- Advertising_Budget: Amount spent on advertising (in thousands of dollars)
- Sales_Revenue: Company's monthly sales revenue (in thousands of dollars)

```python
advertising_budget = np.round(np.random.uniform(5, 50, 24), 2)
sales_revenue = np.round(advertising_budget * np.random.uniform(8, 15, 24), 2)

advertising_sales_data = pd.DataFrame({
    'Advertising_Budget': advertising_budget,
    'Sales_Revenue': sales_revenue
})
advertising_sales_data

plt.figure(figsize=(8,5))
plt.scatter(advertising_sales_data['Advertising_Budget'], advertising_sales_data['Sales_Revenue'],
            color='tab:green', alpha=0.7)
plt.xlabel('Advertising Budget (thousands)')
plt.ylabel('Sales Revenue (thousands)')
plt.title('Advertising Budget vs Sales Revenue')
plt.grid(True)
plt.tight_layout()
plt.show()
```

## 3. Monthly Sales Data

**Description:** Monthly sales revenue for a retail store over the course of a year.

**Columns:**
- Month: Numeric data representing months (1 to 12)
- Sales: Sales in dollars

```python
month = np.arange(1, 13)
sales = np.round(np.random.uniform(5000, 20000, 12), 2)

sales_data = pd.DataFrame({
    'Month': month,
    'Sales': sales
})
sales_data

plt.figure(figsize=(9,5))
plt.bar(sales_data['Month'], sales_data['Sales'], color='tab:orange')
plt.xlabel('Month')
plt.ylabel('Sales (USD)')
plt.title('Monthly Sales')
plt.xticks(sales_data['Month'])
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

## 4. Student Performance in Different Subjects

**Description:** Scores of students in three subjects: Math, Science, and English.

**Columns:**
- Math_Score: Student's Math score
- Science_Score: Student's Science score
- English_Score: Student's English score

```python
math_scores = np.round(np.random.normal(75, 10, 100), 2)
science_scores = np.round(np.random.normal(70, 12, 100), 2)
english_scores = np.round(np.random.normal(80, 8, 100), 2)

student_performance_data = pd.DataFrame({
    'Math_Score': math_scores,
    'Science_Score': science_scores,
    'English_Score': english_scores
})
student_performance_data

#average grades if you need to quickly show which subject students are best at on average.
mean_scores = student_performance_data.mean()
plt.figure(figsize=(7,5))
plt.bar(mean_scores.index, mean_scores.values, color=['#4F81BD', '#C0504D', '#9BBB59'])
plt.title("Average Student Performance Scores")
plt.ylabel("Average Score")
plt.ylim(0, 100)
plt.show()

# Shows the spread, median, and outliers — great for comparing items.
plt.figure(figsize=(8,6))
plt.boxplot([student_performance_data["Math_Score"], student_performance_data["Science_Score"], student_performance_data["English_Score"]],
            labels=["Math", "Science", "English"])
plt.title("Distribution of Scores by Subject")
plt.ylabel("Score")
plt.grid(True)
plt.show()
```

## 5. Stock Prices of Three Companies Over Time

**Description:** Daily closing stock prices of three companies over a year.

**Columns:**
- Day: Days (1 to 365)
- Company_A_Price, Company_B_Price, Company_C_Price: Daily closing prices

```python
days = np.arange(1, 366)
company_a_price = np.round(100 + np.cumsum(np.random.normal(0, 2, 365)), 2)
company_b_price = np.round(150 + np.cumsum(np.random.normal(0, 2.5, 365)), 2)
company_c_price = np.round(200 + np.cumsum(np.random.normal(0, 3, 365)), 2)

stock_data = pd.DataFrame({
    'Day': days,
    'Company_A_Price': company_a_price,
    'Company_B_Price': company_b_price,
    'Company_C_Price': company_c_price
})
stock_data

plt.figure(figsize=(12,5))
plt.plot(stock_data['Day'], stock_data['Company_A_Price'], label='Company A', color='tab:blue')
plt.plot(stock_data['Day'], stock_data['Company_B_Price'], label='Company B', color='tab:orange')
plt.plot(stock_data['Day'], stock_data['Company_C_Price'], label='Company C', color='tab:green')
plt.xlabel('Day')
plt.ylabel('Price')
plt.title('Daily Closing Prices: Company A / B / C')
plt.legend(loc='best')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

## 6. Customer Survey Ratings

**Description:** Ratings given by customers in a survey about the quality of service at a restaurant.

**Columns:**
- Rating: Categorical values (Excellent, Good, Average, Poor, Terrible)

```python
ratings = np.random.choice(
    ['Excellent', 'Good', 'Average', 'Poor', 'Terrible'],
    size=500,
    replace=True,
    p=[0.3, 0.4, 0.2, 0.07, 0.03]
)
survey_data = pd.DataFrame({'Rating': ratings})
survey_data

rating_counts = survey_data['Rating'].value_counts()
labels = rating_counts.index.tolist()
sizes = rating_counts.values
colors = ['#2ca02c', '#1f77b4', '#ff7f0e', '#d62728', '#9467bd']
plt.figure(figsize=(6,6))
plt.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=140, explode=[0.05]*len(labels))
plt.title('Customer Survey Ratings Distribution')
plt.axis('equal')
plt.show()
```

## 7. Distribution of Exam Scores for a University Class

**Description:** Final exam scores of 200 students in a university course.

**Columns:**
- Exam_Score: Numeric exam scores

```python
exam_scores = np.round(np.random.normal(75, 10, 200), 2)

exam_score_data = pd.DataFrame({'Exam_Score': exam_scores})
exam_score_data

plt.figure(figsize=(8,5))
plt.hist(exam_score_data['Exam_Score'], bins=15, color='tab:purple', edgecolor='black', alpha=0.75)
plt.xlabel('Exam Score')
plt.ylabel('Frequency')
plt.title('Distribution of Exam Scores (n=200)')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
